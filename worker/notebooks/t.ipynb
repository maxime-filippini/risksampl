```python
import json
import polars as pl
from worker.database import connect
```

```python
with connect() as conn:
    df_insts = pl.read_database("SELECT * FROM instruments", conn)

df_csv = pl.read_csv("../data/market_data.csv")
```

```python
df_with_tickers = df_csv.join(
    df_insts.with_columns(id=pl.col("id").map_elements(str, return_dtype=pl.String)),
    left_on="instrument_id",
    right_on="id",
).drop("name", "currency", "instrument_id")
```

```python
df_with_tickers.write_csv("../data/market_data_clean.csv")
```
