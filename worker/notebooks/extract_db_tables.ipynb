```python
# Extract data tables to update PROD

from worker.database import connect
import polars as pl
```

```python
tables = [
    "instruments",
    "portfolios",
    "investments",
    "measures",
    "ptf_comp",
    "ptf_values",
    "market_data",
]

dfs = {}

with connect() as session:
    for table in tables:
        df = (
            pl.read_database(f"SELECT * FROM {table}", session)
            .with_columns(pl.selectors.object().map_elements(lambda item: str(item), return_dtype=pl.String))
        )
        dfs[table] = df
```

```python
for k, v in dfs.items():
    v.write_csv(f"../data/db/{k}.csv")
```
